```python
import pandas as pd
```

```python
df_train = pd.read_csv("preprocessed/train_with_bb_features.csv")
df_test = pd.read_csv("preprocessed/test_with_bb_features.csv")
```

```python
df_train.head()
```

```python
categorical_columns = [
    "browser",
    "os",
    "country_id",
    "region_id",
    "timezone_id",
    "domain",
    "event_count",
]
component_columns = [f"component{i}" for i in range(10)] + [
    f"bb_{i}" for i in range(32)
]

df_train[categorical_columns] = df_train[categorical_columns].fillna("NaN").astype(str)
df_test[categorical_columns] = df_test[categorical_columns].fillna("NaN").astype(str)

df_train_x = df_train[categorical_columns + component_columns].copy()
df_train_y = df_train["target"].astype("float32").values

df_test_x = df_test[categorical_columns + component_columns].copy()
df_test_ids = df_test["user_id"].copy()
```

```python
df_train_x.head()
```

```python
df_test_x.head()
```

```python
from sklearn.model_selection import train_test_split

df_train_x, df_val_x, df_train_y, df_val_y = train_test_split(
    df_train_x, df_train_y, test_size=0.1, random_state=42, stratify=df_train_y
)

df_train_x = df_train_x.reset_index(drop=True)
df_val_x = df_val_x.reset_index(drop=True)
```

```python
from category_encoders import CatBoostEncoder

cbe = CatBoostEncoder(cols=categorical_columns, random_state=42)

cbe.fit(df_train_x, df_train_y)

df_train_x_te = cbe.transform(df_train_x)
df_val_x_te = cbe.transform(df_val_x)
df_test_x_te = cbe.transform(df_test_x)
```

```python
df_train_x_te.head()
```

```python
from random import randint
from catboost import CatBoostClassifier


def train_catboost(x_train, y_train, x_val, y_val, seed):
    model = CatBoostClassifier(
        loss_function="Logloss",
        eval_metric="AUC",
        task_type="GPU",
        depth=10,
        l2_leaf_reg=10,
        min_data_in_leaf=100,
        learning_rate=0.01,
        iterations=1000,
        early_stopping_rounds=100,
        random_seed=seed,
        verbose=False,
    )
    model.fit(
        x_train,
        y_train,
        eval_set=(x_val, y_val),
        cat_features=categorical_columns,
        verbose=100,
    )
    return model
```

```python
from tqdm import tqdm

seeds = [randint(1, 10000) for i in range(3)]
models = []

for s in tqdm(seeds):
    print(f"Training model seed={s}")
    m = train_catboost(df_train_x, df_train_y, df_val_x, df_val_y, seed=s)
    models.append(m)
```

```python
import numpy as np

val_probs = np.zeros(len(df_val_y))

for m in models:
    val_probs += m.predict_proba(df_val_x)[:, 1]

val_probs /= len(models)
```

```python
import numpy as np
from sklearn.metrics import accuracy_score


def find_best_threshold(probs, y_true):
    thresholds = np.linspace(0.01, 0.99, 200)
    best_threshold = 0
    best_accuracy = 0

    for thr in thresholds:
        preds = (probs >= thr).astype(int)
        acc = accuracy_score(y_true, preds)
        if acc > best_accuracy:
            best_accuracy = acc
            best_threshold = thr

    return best_threshold, best_accuracy
```

```python
best_thr, best_acc = find_best_threshold(val_probs, df_val_y)
print("Best threshold: ", best_thr)
print("Best accuracy: ", best_acc)
```

```python
test_probs = np.zeros(len(df_test_ids))

for m in models:
    test_probs += m.predict_proba(df_test_x)[:, 1]

test_probs /= len(models)
```

```python
for i, m in enumerate(models):
    m.save_model(f"models/best_model{i}.cbm")
```

```python
pd.DataFrame(
    {"user_id;target": df_test_ids.astype(str) + ";" + test_probs.astype(str)}
).to_csv("predictions/catboost_prediction.csv", index=False)
```
